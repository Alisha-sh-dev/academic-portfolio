Task 2: Visualizing Data Distributions with Seaborn

```python
# Task 2: Visualizing Data Distributions with Seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from io import StringIO

# Load the dataset
csv_data = """
Date,Share Name,Open,High,Low,Close,Volume
2023-08-01,XYZ Corp.,50.00,52.25,48.75,51.20,120000
2023-08-02,XYZ Corp.,51.50,53.80,50.10,53.50,135000
2023-08-03,XYZ Corp.,53.60,54.90,52.25,54.70,110000
2023-08-04,XYZ Corp.,54.80,56.20,53.75,55.90,125000
2023-08-05,XYZ Corp.,56.00,57.40,55.50,56.80,105000
2023-08-08,XYZ Corp.,56.90,58.10,55.80,57.60,100000
2023-08-09,XYZ Corp.,57.70,59.20,57.20,58.90,95000
2023-08-10,XYZ Corp.,58.80,60.00,58.00,59.70,115000
2023-08-11,XYZ Corp.,59.50,61.00,59.10,60.50,110000
2023-08-12,XYZ Corp.,60.70,62.40,60.00,61.80,120000
2023-08-15,XYZ Corp.,61.90,63.20,61.25,62.70,100000
2023-08-16,XYZ Corp.,62.80,64.50,62.00,63.90,105000
2023-08-17,XYZ Corp.,64.00,65.75,63.50,64.80,115000
2023-08-18,XYZ Corp.,65.20,66.80,64.75,66.30,130000
2023-08-19,XYZ Corp.,66.50,68.00,66.25,67.50,125000
2023-08-22,XYZ Corp.,67.80,69.20,67.00,68.60,110000
2023-08-23,XYZ Corp.,68.70,70.00,68.25,69.40,100000
2023-08-24,XYZ Corp.,69.50,71.00,69.10,70.80,105000
2023-08-25,XYZ Corp.,70.90,72.40,70.50,71.70,95000
2023-08-26,XYZ Corp.,71.80,73.20,71.25,72.90,115000
2023-08-29,XYZ Corp.,73.00,75.00,72.75,74.50,120000
2023-08-30,XYZ Corp.,74.80,76.25,74.00,75.60,105000
2023-08-31,XYZ Corp.,75.70,77.00,75.50,76.80,110000
"""

# Create DataFrame
df = pd.read_csv(StringIO(csv_data))
df["Date"] = pd.to_datetime(df["Date"])

# Set Seaborn theme
sns.set(style="whitegrid")

# 1. Distribution Plot: Close Price
plt.figure(figsize=(10, 5))
sns.histplot(df["Close"], bins=10, kde=True, color="skyblue")
plt.title("Distribution of Close Prices")
plt.xlabel("Close Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 2. Box Plot: Price Range (Open, High, Low, Close)
plt.figure(figsize=(10, 6))
sns.boxplot(data=df[["Open", "High", "Low", "Close"]])
plt.title("Box Plot of Stock Prices")
plt.xlabel("Price Type")
plt.ylabel("Price")
plt.tight_layout()
plt.show()

# 3. Pair Plot: Relationships between Price Columns
sns.pairplot(df[["Open", "High", "Low", "Close"]], diag_kind="kde", corner=True)
plt.suptitle("Pairwise Relationships Between Stock Prices", y=1.02)
plt.tight_layout()
plt.show()

# 4. Correlation Heatmap
plt.figure(figsize=(8, 6))
corr_matrix = df[["Open", "High", "Low", "Close", "Volume"]].corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

Task 3: Geographic Data Visualization with GeoPandas

```python
import plotly.express as px

# Load built-in Canadian election data and geographic boundaries
df = px.data.election()
geojson = px.data.election_geojson()

# Visualize vote percentage for candidate 'Coderre'
fig = px.choropleth(
    df,
    geojson=geojson,
    locations="district",                           # Column in df
    featureidkey="properties.district",             # Key in geojson
    color="Coderre",                                # Existing attribute in df
    hover_name="district",                          # Hover info
    hover_data=["Coderre", "Bergeron", "Joly", "winner"],  # Valid columns
    color_continuous_scale="Blues"                  # Custom color scheme
)

# Customize the map layout
fig.update_geos(fitbounds="locations", visible=False)
fig.update_layout(
    title_text="Choropleth Map: Votes for Coderre by District (Canada Election)",
    title_x=0.5,
    margin={"r":0, "t":50, "l":0, "b":0},
    coloraxis_colorbar=dict(title="Votes for Coderre (%)")
)

fig.show()
```
