```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import os

# Define the path to your CSV file
file_path = '/content/drive/My Drive/Asst-3 data.csv'  # Adjust the path if your file is in a different folder

# Read the CSV file into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows of the DataFrame
print(df.head())
```

```python
# Inspect the column names
print("Column names before stripping:", df.columns)
```

```python
# Assuming the data format is consistent, we can split the first column into multiple columns
df[['Date', 'Share Name', 'Open', 'High', 'Low', 'Close', 'Volume']] = df['Date,Share Name,Open,High,Low,Close,Volume'].str.split(',', expand=True)
```

```python
# Parse Date Column
df['Date'] = pd.to_datetime(df['Date'])
```

# Task#1-Line Chart

```python
# Create the Line Plot
plt.figure(figsize=(10, 6))
plt.plot(df['Date'], df['Close'], marker='o', color='b', linestyle='-')

# Step 5: Add Labels and Title
plt.title('Closing Prices Over Time for XYZ Corp.')
plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.grid(True)
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.tight_layout()

# Display the Plot
plt.show()
```

```python
# Import necessary libraries
import geopandas as gpd
import matplotlib.pyplot as plt

# Load the shapefile
shapefile_path = '/content/drive/MyDrive/10m_cultural/ne_10m_admin_0_antarctic_claim_limit_lines.shp'
# Update with your shapefile path
gdf = gpd.read_file(shapefile_path)

# Check the first few rows of the GeoDataFrame
print(gdf.head())
```

```python
print(gdf.columns)
```

```python
# Generate synthetic population density data
import numpy as np

# Assuming 'gdf' is the GeoDataFrame loaded from the shapefile
gdf['population_density'] = np.random.randint(10, 1000, size=len(gdf))  # Generate random population density values

# Plot the choropleth map
fig, ax = plt.subplots(1, 1, figsize=(10, 8))
gdf.plot(column='population_density', cmap='YlGnBu', linewidth=0.8, ax=ax, edgecolor='0.8', legend=True)

# Add title
ax.set_title('Population Density Choropleth Map')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')

# Create and customize legend
cbar = fig.colorbar(ax.collections[0], ax=ax, orientation='vertical')
cbar.set_label('Population Density')

# Show the plot
plt.show()
```
