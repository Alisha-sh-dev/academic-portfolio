```python
# Q1: Speed Calculator with Error Handling
import math

try:
    # Ask user for input
    distance = float(input("Enter the distance traveled in kilometers: "))
    time = float(input("Enter the travel time in seconds: "))

    # Calculate speed
    speed = distance / time

    # Print result
    print(f"Speed = {speed} km/s")

except ZeroDivisionError:
    print("Error: cannot calculate the speed when time is zero.")
except ValueError:
    print("Error: please enter numeric values only.")
```

```python
# Q2: Aircraft Velocity Components
import math

# Given values
speed = 240  # m/s
angle_degrees = 10  # degrees

# Convert angle to radians
angle_radians = math.radians(angle_degrees)

# Calculate velocity components
vx = speed * math.cos(angle_radians)  # horizontal velocity
vz = speed * math.sin(angle_radians)  # vertical velocity

# Print results rounded to 1 decimal place
print(f"Horizontal velocity vx = {vx:.1f} m/s")
print(f"Vertical velocity vz = {vz:.1f} m/s")
```

```python
# Q2: Aircraft Velocity Components with verification
import math

speed = 240  # m/s
angle_degrees = 10  # degrees

angle_radians = math.radians(angle_degrees)
vx = speed * math.cos(angle_radians)
vz = speed * math.sin(angle_radians)

print(f"Aircraft Speed: {speed} m/s")
print(f"Climb Angle: {angle_degrees}°")
print("-" * 35)
print(f"Horizontal velocity vx = {vx:.1f} m/s")
print(f"Vertical velocity vz = {vz:.1f} m/s")

# Verify using Pythagorean theorem
resultant = math.sqrt(vx**2 + vz**2)
print(f"\nVerification: sqrt(vx² + vz²) = {resultant:.1f} m/s (should be {speed})")
```

```python
# Q3: Dynamic Pressure Calculation using NumPy
import numpy as np

# Create arrays
rho = np.array([1.225, 1.18, 1.10, 1.00])
v = np.array([70, 110, 150, 200])

# Calculate dynamic pressure
q = 0.5 * rho * v**2

# Print result rounded to 1 decimal place
print("Dynamic Pressure Values:")
print(np.round(q, 1))
```

```python
# Q4: 2D Telemetry Array Analysis
import numpy as np

# Create 2D array
telemetry = np.array([
    [0, 0, 0],
    [5, 160, 48],
    [10, 510, 95],
    [15, 980, 140]
])

# Extract altitude column (column index 1)
altitude = telemetry[:, 1]

# Calculate mean altitude
mean_altitude = np.mean(altitude)

# Print results
print("Altitude values:", altitude)
print("Mean altitude:", mean_altitude)
```
