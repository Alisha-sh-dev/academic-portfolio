Q1. Create and reshape NumPy arrays

```python
import numpy as np

# a. Create a vector of four zeros
zeros_vec = np.zeros(4)
print("The array of zeros is:", zeros_vec)

# b. Create a vector of four ones
ones_vec = np.ones(4)
print("The array of ones is:", ones_vec)

# c. Create a vector of six random integers between 1 and 9 (inclusive)
random_vec = np.random.randint(1, 10, 6)
print("The random array is:", random_vec)

# d. Reshape array [1,2,3,4,5,6] into a 2x3 matrix
arr = np.array([1, 2, 3, 4, 5, 6])
reshaped = arr.reshape(2, 3)
print("The reshaped matrix is:")
print(reshaped)
```

Q2. Vectors and matrix operations

```python
import numpy as np

# Create vectors and matrix
a = np.array([2, 3])
b = np.array([5, 1])
M = np.array([[1, 2], [0, 1]])

# Sum of vectors a and b
sum_ab = a + b
print("The sum of a and b is:", sum_ab)

# Inner (dot) product of a and b
dot_product = np.dot(a, b)
print("The inner product of a and b is:", dot_product)

# Matrix product M * b
Mb = np.dot(M, b)
print("The matrix product Mb is:", Mb)
```

Q3. Solve a linear system

```python
import numpy as np

# Solve the system:
# 3x + y = 9
# x + 2y = 8

# Coefficient matrix A and constant vector b
A = np.array([[3, 1], [1, 2]])
b = np.array([9, 8])

# Solve the linear system
solution = np.linalg.solve(A, b)
x, y = solution

print(f"The solution is:")
print(f"x = {x:.2f}, y = {y:.2f}")
```

```python
from scipy.optimize import root_scalar

# Define the function
def f(x):
    return x**2 - 9

# Find positive root in interval [0, 5]
result = root_scalar(f, bracket=[0, 5], method='bisect')
root = result.root

print(f"The positive root is: {root:.1f}")
```

Q5. Integration from data

```python
import numpy as np

# Data points
time = np.array([0, 1, 2, 3, 4])
speed = np.array([0, 4, 7, 9, 10])

# Estimate area under the speed-time graph using trapezoidal rule
area = np.trapezoid(speed, time)

print(f"The estimated area under the graph is: {area:.1f}")
```

Q6. Differential equations using the Euler method

```python
import numpy as np

# Given values
T = 90
T_env = 25
k = 0.02
dt = 1
steps = 10

# Euler method
for i in range(steps):
    dT = -k * (T - T_env)
    T = T + dT * dt

# Round to 2 decimal places
print(f"The final value of T after {steps} steps is: {T:.2f}")
```
