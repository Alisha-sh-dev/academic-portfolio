```python
# Exercise 1: Count lines that start with "Project"
# Open and read the file
file = open("engineering_project_log.txt", "r")

# Initialize counter
count = 0

# Read file line by line
for line in file:
    # Check if line starts with "Project" (case-sensitive)
    if line.startswith("Project"):
        count += 1

# Close the file
file.close()

# Print result
print(f"There are {count} lines that start with 'Project'")
```

```python
# Exercise 2: Read CSV and calculate totals
with open("student_scores.csv", "r") as file:
    # Skip the first line (header)
    header = file.readline()

    # Read and process each student record
    for line in file:
        # Remove newline and split by comma
        parts = line.strip().split(",")

        # Extract data
        name = parts[0]
        quiz = int(parts[1])
        midterm = int(parts[2])
        final = int(parts[3])

        # Calculate total
        total = quiz + midterm + final

        # Print result
        print(f"{name} {total}")
```
