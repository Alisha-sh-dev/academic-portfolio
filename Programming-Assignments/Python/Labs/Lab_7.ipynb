```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values from 1 to 100 inclusive
x = np.arange(1, 101)

# Calculate y = 1 + 2x + x²
y = 1 + 2*x + x**2

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, color='blue', linewidth=2)
plt.title('Quadratic Function: y = 1 + 2x + x²', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

Q2. Draw a bar chart using matplotlib

```python
import matplotlib.pyplot as plt

# Student data
students = ['Ali', 'Sara', 'John', 'Lina', 'Omar']
marks = [78, 85, 92, 88, 95]

# Create the bar chart
plt.figure(figsize=(8, 6))
bars = plt.bar(students, marks, color='skyblue', edgecolor='black')

# Add value labels on top of bars
for bar, mark in zip(bars, marks):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1, 
             str(mark), ha='center', va='bottom', fontsize=12, fontweight='bold')

plt.title('Student Marks', fontsize=14)
plt.xlabel('Students', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.ylim(0, 100)
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

Q3. Create a DataFrame and retrieve data using iloc

```python
import pandas as pd

# Create the DataFrame
data = {
    'Name': ['Ali', 'Sara', 'John', 'Lina', 'Omar'],
    'Age': [20, 21, 19, 22, 20],
    'Mark': [78, 85, 92, 88, 95],
    'City': ['Doha', 'Cairo', 'Lagos', 'Dubai', 'Muscat']
}

df = pd.DataFrame(data)

# Display rows 1 to 3 and columns 0 to 1 using iloc
# Note: iloc uses 0-based indexing, so rows 1-3 means indices 1,2,3
# Columns 0-1 means columns 0 and 1
result = df.iloc[1:4, 0:2]
print(result)
```

Q4. Select columns and save the result as CSV

```python
import pandas as pd

# Create the DataFrame (same as Q3)
data = {
    'Name': ['Ali', 'Sara', 'John', 'Lina', 'Omar'],
    'Age': [20, 21, 19, 22, 20],
    'Mark': [78, 85, 92, 88, 95],
    'City': ['Doha', 'Cairo', 'Lagos', 'Dubai', 'Muscat']
}

df = pd.DataFrame(data)

# Select only Name and Mark columns
selected_data = df[['Name', 'Mark']]

# Display the selected result on screen
print("Selected Data:")
print(selected_data)

# Save the result to a CSV file
selected_data.to_csv('student_marks.csv', index=False)

# Verify the CSV file content
print("\nCSV file saved as 'student_marks.csv'")
```
