```python
# Program 3: Print all odd numbers from 1 to 99
print("Odd numbers from 1 to 99:")
for num in range(1, 100, 2):
    print(num)
```

```python
# Program 4: Square calculator using while loop
while True:
    num = float(input("Please enter a nonzero number, and I will print its square for you. Enter 0 if you want me to stop: "))
    
    if num == 0:
        print("The program is now finished. Thank you.")
        break
    else:
        square = num ** 2
        print(f"The square of {num} is {square}")
```

```python
# Program 5: Find odd roots of equation x^3 - 24x^2 - 81x = 0
# Equation: x^3 - 24x^2 - 81x = 0
# Factor: x(x^2 - 24x - 81) = 0
# Roots: x = 0, x = 27, x = -3 (but we only check 1 to 99)

found = False
for x in range(1, 100):
    # If number is even, skip
    if x % 2 == 0:
        continue
    
    # Check if x is a root of the equation
    result = x**3 - 24*x**2 - 81*x
    
    # If x is a root and odd, print and break
    if result == 0:
        print(f"{x} is an odd root found!")
        found = True
        break
    else:
        print(x)

if not found:
    print("No odd roots found in range 1 to 99")
print("finished")
```
