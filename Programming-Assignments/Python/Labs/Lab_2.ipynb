```python
# Program 1: Thermal Camera Temperature Monitoring
temps = [70, 72, 75, 78, 82, 85, 79, 74]  # °C

# Extract left and right zones
left_zone = temps[0:4]   # First 4 values
right_zone = temps[4:8]  # Last 4 values

# Check for overheating
left_overheating = False
right_overheating = False

for temp in left_zone:
    if temp > 80:
        left_overheating = True
        break

for temp in right_zone:
    if temp > 80:
        right_overheating = True
        break

# Print results
if left_overheating:
    print("Left overheating")
elif right_overheating:
    print("Right overheating")
else:
    print("All zones safe")
```

```python
# Program 2: Material Yield Strength Dictionary
materials = {
    "Aluminum": 270,
    "Steel": 400,
    "Titanium": 880
}

# Ask user for material name
material_name = input("Enter material name: ")

# Check if material exists in dictionary
if material_name in materials:
    print(f"Strength of {material_name}: {materials[material_name]} MPa")
else:
    print("Material not found")
```

```python
# Program 3: Aircraft Telemetry Analysis
code = "GPS TO00"

# Extract phase (positions 4-5, after the space)
phase = code[4:6]  # "TO"

# Determine phase type
if phase == "TO":
    print("Takeoff phase")
elif phase == "CL":
    print("Climb phase")
elif phase == "CR":
    print("Cruise phase")
else:
    print("Unknown phase")
```

```python
# Program 4: Satellite Battery Status Check
batteries = {
    "BAT1": "OK",
    "BAT2": "LOW",
    "BAT3": "OK",
    "BAT4": "OK"
}

# Check overall status
has_failed = False
has_low = False

for status in batteries.values():
    if status == "FAILED":
        has_failed = True
        break
    elif status == "LOW":
        has_low = True

# Determine and print status
if has_failed:
    print("Critical failure")
elif has_low:
    print("Power warning")
else:
    print("Power stable")
```

```python
# Program 5: Sensor Data Packet Analysis
packet = "T235P1013H55"

# Extract data using slicing
temp_str = packet[1:4]     # "235"
pressure_str = packet[5:9] # "1013"
humidity_str = packet[10:12] # "55" (optional)

# Convert to numbers
temp = int(temp_str) / 10   # Convert to actual temperature: 23.5°C
pressure = int(pressure_str)  # 1013 hPa

# Print decoded values
print(f"Temperature: {temp}°C")
print(f"Pressure: {pressure} hPa")
print(f"Humidity: {humidity_str}%")

# Determine status
if temp > 50:
    print("High temp alert")
elif pressure < 1000:
    print("Low pressure")
else:
    print("Readings normal")
```
