```python
# Program 1

def temperature_celsius(altitude_m):
    # Check if altitude is within valid range
    if altitude_m < 0 or altitude_m > 11000:
        return None

    # Temperature decreases with altitude
    temperature = 15.0 - (0.0065 * altitude_m)

    return temperature


# Test calls
print(temperature_celsius(0))
print(temperature_celsius(5000))
print(temperature_celsius(11000))
print(temperature_celsius(12000))
```

```python
# Program 2

import math


def temperature_celsius(altitude_m):
    # Check altitude range
    if altitude_m < 0 or altitude_m > 11000:
        return None

    return 15.0 - (0.0065 * altitude_m)


def mach_number(speed_mps, altitude_m):

    # Speed must be positive
    if speed_mps <= 0:
        return None

    # Get temperature at altitude
    temp_c = temperature_celsius(altitude_m)

    # Invalid altitude
    if temp_c is None:
        return None

    # Convert Celsius to Kelvin
    temp_k = temp_c + 273.15

    # Speed of sound formula
    speed_of_sound = math.sqrt(1.4 * 287 * temp_k)

    # Calculate Mach number
    mach = speed_mps / speed_of_sound

    return round(mach, 3)


# Required test cases

# Sea level
print(mach_number(340, 0))

# High altitude
print(mach_number(250, 10000))

# Normal cases
print(mach_number(220, 0))
print(mach_number(300, 5000))

# Invalid speed
print(mach_number(-100, 5000))

# Invalid altitude
print(mach_number(250, -500))
print(mach_number(250, 12000))
```

```python
# Program 3

import math


def temperature_celsius(altitude_m):
    if altitude_m < 0 or altitude_m > 11000:
        return None

    return 15.0 - (0.0065 * altitude_m)


def mach_number(speed_mps, altitude_m):

    if speed_mps <= 0:
        return None

    temp_c = temperature_celsius(altitude_m)

    if temp_c is None:
        return None

    temp_k = temp_c + 273.15

    speed_of_sound = math.sqrt(1.4 * 287 * temp_k)

    mach = speed_mps / speed_of_sound

    return round(mach, 3)


# Given list
speeds_mps = [120, 180, 220, 250, 300, 340]

# List comprehension
mach_values = [mach_number(speed, 0) for speed in speeds_mps]

print(mach_values)
```
